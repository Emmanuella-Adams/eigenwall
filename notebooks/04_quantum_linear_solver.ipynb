# 04 - HHL-Style Quantum Linear System Solver
Solves $Ax = b$ using QPE, controlled eigenvalue inversion rotation, and uncomputation.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
from eigenwall.quantum.hhl_solver import quantum_linear_solver
from eigenwall.metrics import state_fidelity

A = np.array([[1.0, 0.2], [0.2, 2.0]], dtype=complex)
b = np.array([1.0, 1.0], dtype=complex)
res = quantum_linear_solver(A, b, phase_qubits=4, shots=4096)
print('Classical solution state:', res['classical_state'])
print('Quantum solution state:', res['solution_state'])
print('Fidelity F:', res['fidelity'])
print('P(success):', res['success_probability'])
```
