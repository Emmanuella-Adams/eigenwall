# 03 - Quantum Phase Estimation (QPE)
Verifies unitary evolution $U = e^{iRt}$ and eigenvalue estimation on analytically known benchmark matrices.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
from eigenwall.quantum.qpe import run_qpe

R1 = np.array([[1.0, 0.0], [0.0, 2.0]], dtype=complex)
counts, res = run_qpe(R1, t_val=np.pi/2.0, m=4, input_state_vector=np.array([1.0, 0.0]), lambda_true=1.0)
print('Dominant bitstring:', res['bitstring'])
print('Estimated eigenvalue:', res['lambda_est'])
print('Quantization error bound:', res['max_q_error_lambda'])
```
