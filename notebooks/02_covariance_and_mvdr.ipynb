# 02 - Covariance Estimation & Classical MVDR Beamforming
Estimates sample covariance matrix $R$, verifies Hermiticity & positive definiteness, and computes classical MVDR weights.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
import matplotlib.pyplot as plt
from eigenwall.signal_model import ArrayConfig, steering_vector, generate_received_signal, generate_sinusoid
from eigenwall.covariance import sample_covariance, regularize_covariance, check_covariance_properties
from eigenwall.beamforming import mvdr_weights, beam_pattern

config = ArrayConfig(n_elements=4)
s = generate_sinusoid(5.0, 1000)
i = generate_sinusoid(18.0, 1000, amplitude=1.8)
X = generate_received_signal(config, 20.0, s, [(-30.0, i)], noise_power=0.8, random_seed=42)

R = regularize_covariance(sample_covariance(X), epsilon=1.0)
props = check_covariance_properties(R)
print('Condition number:', props['condition_number'])

a_target = steering_vector(4, 20.0)
w_mvdr = mvdr_weights(R, a_target)
print('MVDR weights:', w_mvdr)
```
